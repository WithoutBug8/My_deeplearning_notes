# Zero-shot Learning（零样本学习）

## 定义

Zero-shot Learning（零样本学习）是指**模型在训练阶段没有见过某一类别或任务的样本，但能够利用已有知识完成对新类别或新任务的推理。**

## 在 Prompt Engineering（提示词工程）中的应用

**Zero-shot Prompting** 是指**不给模型提供任何示例（Example），仅通过自然语言描述任务要求、目标和约束，让模型直接完成任务。**

### 特点

- 不提供任何示例（Example）
- 完全依赖模型已有知识和推理能力
- Prompt 中只有任务描述

### 示例

**Prompt：**

```text
请将下面这段新闻总结为一句话。
```

模型无需参考任何示例，直接根据指令完成任务。

### 传统 Zero-shot Learning 示例

训练时模型没有见过"斑马"这一类别，但已知：

- 马：四条腿
- 黑白条纹：一种外观特征

推理阶段告诉模型：

> 斑马是一种具有黑白条纹的马。

模型便可以根据已有知识推理出"斑马"这一新类别。

# Few-shot Learning（少样本学习）

## 定义

Few-shot Learning（少样本学习）是指**模型已经具备丰富的基础知识，对于新的任务或类别，只需要提供少量样本（Few Examples），即可完成推理或学习。**

## 在 Prompt Engineering（提示词工程）中的应用

**Few-shot Prompting** 是指**在 Prompt 中提供少量输入-输出示例（Examples），让模型学习任务格式和推理方式，再完成新的任务。**

### 特点

- 提供少量示例（通常 2～5 个）
- 帮助模型理解任务要求和输出格式
- 适用于格式固定、复杂任务或需要统一输出风格的场景

### 示例

任务：抽取产品名称和核心卖点，输出 JSON。

**示例 1**

输入：

```text
MacBook Pro 高效节能，性能强大，适合办公使用。
```

输出：

```json
{
  "产品名称": "MacBook Pro",
  "产品卖点": "高效节能，性能强大"
}
```

---

**示例 2**

输入：

```text
联想笔记本拥有 RTX4060 独立显卡，畅玩游戏，丝滑流畅。
```

输出：

```json
{
  "产品名称": "联想笔记本",
  "产品卖点": "畅玩游戏，丝滑流畅"
}
```
