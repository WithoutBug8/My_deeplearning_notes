# 张量的运算函数

张量类似于numpy可以有一些系统自带的统计函数调用
- sum() 求和
- mean() 平均值
- max()/min() 最大最小值
- argmax()/argmin()求最大值/最小值的索引
- std()求标准差
- unique()去重
- sort()排序

```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 2, 4)).float()
print("tensor1:", tensor1)
```

```python
# 求和函数
print("所有元素的综合，结果tensor1.sum():", tensor1.sum())

# 不同维度的求和
print("在维度1上求和，结果tensor1.sum(dim=1):", tensor1.sum(dim=1))
```

```python
# 求均值
print("所有元素的均值，结果tensor1.mean():", tensor1.mean())
print("在维度0上求均值，结果tensor1.mean(dim=0):", tensor1.mean(dim=0))
```

```python
# 求标准差
print("所有元素的标准差，结果tensor1.std():", tensor1.std())
print("在维度2上求标准差，结果tensor1.std(dim=2):", tensor1.std(dim=2))
```

```python
# 求最大和最小
print("所有元素的最大值，结果tensor1.max():", tensor1.max())
print("所有元素的最小值，结果tensor1.min():", tensor1.min())
# 某一个维度的最大值和最小值
'''
 dim= n 表示在第n维上比较两个数据的大小，比如dim=1 就是行之间比较，dim=2就是列之间比较。
'''
print("在维度1上求最大值，结果tensor1.max(dim=1):", tensor1.max(dim=1))
print("在维度1上求最小值，结果tensor1.min(dim=1):", tensor1.min(dim=1))
```

```python
# 找最大值和最小值的索引
print("在维度1上求最大值的索引，结果tensor1.argmax(dim=1):", tensor1.argmax(dim=1))
print("在维度1上求最小值的索引，结果tensor1.argmin(dim=1):", tensor1.argmin(dim=1))
```

```python
# 去重
print("tensor1去重后的结果torch.unique(tensor1):", tensor1.unique())
```

```python
# 排序，默认是按照dim=2排序
print("tensor1排序后的结果torch.sort(tensor1):", torch.sort(tensor1))
```
