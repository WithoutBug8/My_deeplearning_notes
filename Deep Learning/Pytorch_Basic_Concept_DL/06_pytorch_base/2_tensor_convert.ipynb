# 第二节 张量的转换

```python
import numpy as np
import torch
```

### 张量的类型转换

```python
tensor1 = torch.tensor([1,2,3])
print('转换前的类型:',tensor1.dtype)

tensor1 = tensor1.type(torch.float32)
print('转换后的类型:',tensor1.dtype)

# PyTorch有一个特殊的数据类型，复数类型(实数虚数合在一起的类型)，torch.complex64和torch.complex128
tensor1 = tensor1.to(torch.complex64)
print(tensor1)
print('转换成复数类型的:',tensor1.dtype)
```

### 1.Tensor和ndarray之间相互转换
- 原理是共享内存：就是**数据存放在内存的某一个区间**，然后ndarray指向这部分内存中的数据，并且tensor也是指向这部分内存的数据
    - 共享内存可能会导致，本来tensor不想修改数据，但是ndarray修改完成了数据，所以为了避免数据不一致，会采用`copy()`来避免共享内存造成的问题，缺点是占用更多的内存

```python
np.set_printoptions(precision=2) # 设置numpy的打印精度
torch.set_printoptions(precision=2) # 设置torch的打印精度
```

```python
# Tensor和ndarray相互转换
tensor1 = torch.rand(3,2)
print('tensor1转换之前的样子:',tensor1)
# 转换之后精度不一样，可以设置精度
ndarray1 = tensor1.numpy()
print('tensor1转换成ndarray的样子:',ndarray1)
```

```python
# 修改ndarray1的值，tensor1的值也会改变
tensor1[:,0] = 10
print('修改后tensor1的值:\n',tensor1)
print('ndarray1的值也会随之改变:\n',ndarray1)
```

```python
# 使用Copy方法，修改ndarray1的值，tensor1的值不会改变
tensor1 = torch.rand(3,2)
print('tensor1转换之前的样子:\n',tensor1)
ndarray1 = tensor1.numpy()
print('ndarray1的样子:',ndarray1)
ndarray2 = tensor1.numpy().copy()
print('ndarray2的样子:\n',ndarray2)
tensor1[:,0] = 5
print('修改后tensor1的值:\n',tensor1)
print('ndarray1的值也会随之改变:\n',ndarray1)
print('ndarray2的值不会随之改变:\n',ndarray2)
```

```python
# ndarray转换成tensor
ndarray1 = np.random.randn(3,2)
tensor1 = torch.from_numpy(ndarray1)
print('ndarray1的样子:\n',ndarray1)
print('ndarray1转换成tensor1的样子:\n',tensor1)
```

```python
# 使用copy方法避免内存共享
tensor2 = torch.from_numpy(ndarray1.copy())

ndarray1[:,0] = 10
print('修改后ndarray1的值:\n',ndarray1)
print('tensor1的值也会随之改变:\n',tensor1)
print('tensor2的值不会随之改变:\n',tensor2)
```

#### 直接将ndarray转换为tensor，不共享内存的
 - 使用`torch.tensor(ndarray)`方法, 不用每次都调用`Copy()`方法，更加简便

```python
ndarray1 = np.random.randn(3,2)
tensor1 = torch.tensor(ndarray1)
print('ndarray1的样子:\n',ndarray1)
print('ndarray1转换成tensor1的样子:\n',tensor1)
```

### 2. 张量Tensor和标量之间的转换
- 标量：只有一个元素
- **转换的前提是：** 张量中只有一个元素

```python
tensor = torch.tensor([10])
print('tensor的样子:\n',tensor)
print('转换为标量的样子:\n',tensor.item())
```
