# 张量的索引操作
- 简单索引
- 列表索引
- 布尔索引

```python
import torch
```

### 简单索引

```python
tensor1 = torch.randint(1, 10, (3,5,4))
print("tensor1:", tensor1)
```

```python
# 简单索引操作
print("给定下标寻找特定元素", tensor1[2,1,3])
print("给定下标寻找特定行", tensor1[2,1])
print("给定下标寻找所有矩阵的第三行", tensor1[:,3])
```

```python
# 范围索引
print("范围索引，寻找第0-1行的元素", tensor1[0:2])
print("取最后一个矩阵", tensor1[-1:])
print("取最后一个矩阵的第1-3列和第0-2行",tensor1[-1,1:4,0:3])
```

### 列表索引

```python
print("取第0和第1行，第1和第2列的元素",tensor1[[0,1],[1,2]])
'''
高级索引的使用方法，看下面的案例：tensor1[[0,1],[3,2],[2,3]]
就是取这两个位置tensor1[0, 3, 2]和tensor1[1, 2, 3]
'''
print(tensor1[[0,1],[3,2],[2,3]])

'''
列表的多个对应关系,广播
第一个列表的0和1要分别对应第二个列表的1和2
tensor1[0, 1, :]
tensor1[0, 2, :]

tensor1[1, 1, :]
tensor1[1, 2, :]
'''

print(tensor1[[[0],[1]],[1,2]])
```

### 布尔索引

```python
# 符合条件的行和列
mask =  tensor1[:,:,0]>5
print("符合条件的行，首元素大于5:mask", mask)
'''
这里打印出来的都是bool值，所以需要把值存到mask中，然后再用tensor1[mask]来打印符合条件的行
注意：tensor1[mask]的形状要和原张量匹配，否则会报错
'''
print("符合条件的行，首元素大于5", tensor1[mask])
```
